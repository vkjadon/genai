```python
from google import genai
from google.colab import userdata

client = genai.Client(api_key=userdata.get('GOOGLE_API_KEY'))
```

```python
model = "gemini-2.5-flash-lite" # @param ["gemini-2.5-flash-lite", "gemini-2.5-flash", "gemini-2.5-pro","gemini-3-pro-preview"] {"allow-input":true, isTemplate: true}
```

```python
from google.genai import types
```

```python
# Start a chat session
# In 2025, the 'gemini-2.0-flash' or 'gemini-2.5-flash' models are perfect for chat
chat_session = client.chats.create(
    model=model,
    config=types.GenerateContentConfig(
        system_instruction="You are a personal travel guide. Keep responses under 3 sentences."
    )
)

# Send the first message
response1 = chat_session.send_message("I am planning a trip to Tokyo in October.")
print(f"AI: {response1.text}")

# Send a follow-up (without repeating that we're talking about Tokyo)
response2 = chat_session.send_message("What should I pack?")
print(f"AI: {response2.text}")
```

```python
response = chat_session.send_message("Where am I going?")
print(f"AI: {response.text}")
```
